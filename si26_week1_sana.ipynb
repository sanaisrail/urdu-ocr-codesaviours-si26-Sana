1. What is OCR (Optical Character Recognition)?

Answer:  Optical Character Recognition (OCR) is a technology that helps computers read text from images, scanned documents, or photos. It converts the text into a digital format that can be edited, copied, or searched. OCR is very useful because it saves time and reduces the need to type everything manually.

2. Why is Urdu OCR harder than English OCR?

Answer: Urdu OCR is more difficult than English OCR because Urdu is written from right to left, and many letters change their shape depending on where they appear in a word. Some Urdu letters also look very similar, with only small differences like dots. Different handwriting styles and unclear images make it even harder for a computer to recognize the text correctly.

3. What are two real-world situations where Urdu OCR would be useful?

Answer:Urdu OCR can help digitize old books, newspapers, and historical documents so that they can be preserved and searched easily in the future. It is also useful for businesses and government offices to convert printed Urdu forms, records, and official documents into editable digital text, reducing manual work and improving efficiency. These applications save time, improve accuracy, and make important information easier to access and manage.

```python
from PIL import Image, ImageDraw, ImageFont
import arabic_reshaper
from bidi.algorithm import get_display
import os
import random

urdu_texts = [
    "پاکستان زندہ باد",
    "آج کا موسم خوشگوار ہے",
    "تعلیم ہر انسان کا حق ہے",
    "کراچی پاکستان کا سب سے بڑا شہر ہے",
    "محنت کامیابی کی کنجی ہے",
    "علم روشنی ہے",
    "وقت بہت قیمتی ہے",
    "محنت کا پھل میٹھا ہوتا ہے",
    "ہم سب پاکستانی ہیں",
    "صاف ماحول صحت کی ضمانت ہے",
    "درخت لگائیں ماحول بچائیں",
    "کتاب بہترین دوست ہے",
    "ہر انسان برابر ہے",
    "سچ ہمیشہ کامیاب ہوتا ہے",
    "پاکستان ایک خوبصورت ملک ہے",
    "اردو ہماری قومی زبان ہے",
    "پڑھائی کامیابی کی پہلی سیڑھی ہے",
    "ہمیں دوسروں کی مدد کرنی چاہیے",
    "دیانت داری بہترین پالیسی ہے",
    "اتحاد ایمان اور نظم",
    "ہمیشہ سچ بولیں",
    "علم انسان کو عظیم بناتا ہے",
    "ماں کی دعا جنت کی ہوا",
    "صفائی نصف ایمان ہے",
    "محبت نفرت سے بہتر ہے",
    "صحت سب سے بڑی دولت ہے",
    "وقت کی قدر کریں",
    "ہر دن ایک نئی امید ہے",
    "کامیابی مسلسل محنت سے ملتی ہے",
    "مثبت سوچ کامیابی کی بنیاد ہے",
    "ہمیں اپنے ملک سے محبت ہے",
    "خوش رہیں اور دوسروں کو خوش رکھیں",
    "امن سب کی ضرورت ہے",
    "صبر اور شکر کامیابی کا راز ہیں",
    "ہر مشکل کے بعد آسانی ہے",
    "دوسروں کا احترام کریں",
    "زندگی ایک خوبصورت نعمت ہے",
    "اچھی عادتیں کامیابی لاتی ہیں",
    "کتابیں علم کا خزانہ ہیں",
    "نیکی کبھی ضائع نہیں ہوتی"
]

backgrounds = [
    (255, 255, 255),
    (245, 245, 220),
    (230, 240, 255),
    (240, 255, 240),
    (255, 250, 205),
    (255, 228, 225),
    (224, 255, 255),
    (250, 235, 215),
    (240, 248, 255),
    (245, 255, 250)
]

text_colors = [
    "black",
    "darkblue",
    "darkgreen",
    "maroon",
    "purple",
    "brown"
]

os.makedirs("data/synthetic", exist_ok=True)

font = ImageFont.truetype(
    "/usr/share/fonts/truetype/noto/NotoNaskhArabic-Regular.ttf",
    32
)

for i, text in enumerate(urdu_texts):
    reshaped = arabic_reshaper.reshape(text)
    bidi_text = get_display(reshaped)

    width = random.randint(650, 800)
    height = random.randint(100, 150)

    bg = random.choice(backgrounds)
    txt_color = random.choice(text_colors)

    img = Image.new("RGB", (width, height), bg)
    draw = ImageDraw.Draw(img)

    bbox = draw.textbbox((0, 0), bidi_text, font=font)
    text_width = bbox[2] - bbox[0]
    text_height = bbox[3] - bbox[1]

    x = width - text_width - 20
    y = (height - text_height) // 2

    draw.text((x, y), bidi_text, fill=txt_color, font=font)

    img.save(f"data/synthetic/urdu_{i+1}.png")

print(f"{len(urdu_texts)} synthetic images generated successfully!")
```

```python
from pathlib import Path
from PIL import Image

# Book folder path
folder = Path("data/raw/book")

# Supported extensions
extensions = ["*.jpg", "*.jpeg", "*.JPG", "*.JPEG", "*.png", "*.PNG"]

for ext in extensions:
    for img_path in folder.glob(ext):
        img = Image.open(img_path).convert("RGB")

        # Save as PNG
        new_path = img_path.with_suffix(".png")
        img.save(new_path, "PNG")

        # Delete old file if it wasn't already PNG
        if img_path.suffix.lower() != ".png":
            img_path.unlink()

print("✅ All images converted to PNG.")
```

```python
import os
import pandas as pd

# Read existing labels.csv
df = pd.read_csv("/content/labels.csv")

folders = [
    ("books", 0, 40),
    ("newspaper", 40, 80),
    ("other", 80, 120),
    ("synthetic", 120, 160),
    ("signboards", 160, 200),
]

for folder, start, end in folders:
    folder_path = f"data/raw/{folder}"

    # Read image files only
    files = sorted([
        f for f in os.listdir(folder_path)
        if f.lower().endswith((".png", ".jpg", ".jpeg"))
    ])

    print(folder, len(files))

    if len(files) != (end - start):
        print(f"⚠️ {folder}: Expected {end-start}, Found {len(files)}")

    # Replace image paths
    for i, file in enumerate(files):
        df.loc[start + i, "image"] = os.path.join(folder_path, file)

# Save new CSV
df.to_csv("/content/labels_fixed.csv", index=False, encoding="utf-8-sig")

print("✅ Saved: /content/labels_fixed.csv")
```

```python
import os

folders = [
    "/content/data/raw/newspaper",
    "/content/data/raw/signboard",
    "/content/data/raw/book",
    "/content/data/raw/other",
    "/content/data/raw/synthetic"
]

for folder in folders:
    if os.path.exists(folder):
        for filename in os.listdir(folder):
            old_path = os.path.join(folder, filename)

            name, ext = os.path.splitext(filename)
            new_filename = name + ext.lower()   # .PNG -> .png, .JPG -> .jpg
            new_path = os.path.join(folder, new_filename)

            if old_path != new_path:
                os.rename(old_path, new_path)

print("✅ All image extensions converted to lowercase.")
```
